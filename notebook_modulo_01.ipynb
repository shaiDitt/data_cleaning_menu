### 6. Deteção de Outliers (IQR)

Identificação de potenciais valores discrepantes na coluna `release_year` utilizando o método do Intervalo Interquartil (IQR). Este cálculo permite-nos definir os limites superior e inferior aceitáveis para os anos de lançamento, isolando os registos que fogem ao padrão.

```python
print(f"## Cálculo dos outliers pelo método IQR ##")
# 1. Cálculo dos quartis e do IQR
q1 = df['release_year'].quantile(0.25)
q3 = df['release_year'].quantile(0.75)
iqr = q3 - q1
 
# 2. Definição dos limites inferior e superior
limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr
 
# 3. Filtrar os registros fora dos limites
outliers_anos = df[(df['release_year'] < limite_inferior) | (df['release_year'] > limite_superior)]
 
# Exibir os resultados
print(f"Q1 (25%): {q1:.0f}")
print(f"Q3 (75%): {q3:.0f}")
print(f"IQR: {iqr:.0f}")
print(f"Limite Inferior: {limite_inferior:.0f}")
print(f"Limite Superior: {limite_superior:.0f}")
print(f"Total de títulos considerados outliers: {len(outliers_anos)} ({len(outliers_anos)/len(df)*100:.2f}%)")
```

### 7. Visualização Gráfica

Nesta secção, utilizamos recursos visuais para ilustrar as tendências, distribuições e padrões identificados na análise dos dados da Netflix.

#### 7.1 Evolução Temporal da Produção de Filmes

Análise de linha focada exclusivamente no tipo `Movie`, demonstrando o crescimento na quantidade de filmes lançados ao longo dos anos.

```python
#Um gráfico simples (histograma, boxplot) sempre que ajude a ilustrar um resultado
print("## Quantidade de filmes feito por ano ##")

# 1. Filtrar apenas títulos do tipo 'Movie' e contar por ano de lançamento
filmes_por_ano = (
    df[df['type'] == 'Movie']['release_year']
    .value_counts()
    .sort_index()
    .reset_index()
)
filmes_por_ano.columns = ['release_year', 'total_filmes']
 
# 2. Configurar o estilo do gráfico
plt.figure(figsize=(12, 5))
sns.set_theme(style="whitegrid")
 
# 3. Criar o gráfico de linha
ax = sns.lineplot(
    data=filmes_por_ano, 
    x='release_year', 
    y='total_filmes', 
    color='#E50914', # Vermelho Netflix
    linewidth=2.5
)
 
# 4. Personalizar títulos e rótulos
plt.title('Evolução do Número de Filmes por Ano de Lançamento', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Ano de Lançamento', fontsize=11)
plt.ylabel('Quantidade de Filmes', fontsize=11)
 
# Ajustar os limites do eixo X para focar no período relevante (ex: a partir de 1940)
plt.xlim(1940, filmes_por_ano['release_year'].max() + 1)
 
plt.tight_layout()
plt.show()
```

### 8. Modificando Dataset

Como o dataset `netflix_titles` não possui valores duplicados e a sua única coluna numérica (`release_year`) já está preenchida, realizamos algumas alterações na base original para testar e validar o menu de tratamento de dados, gerando como resultado o ficheiro `netflix_titles_modificado.csv`.

#### 8.1 Criando Linhas Duplicadas

Como o dataset original não contém registros duplicados, selecionamos alguns intervalos específicos de linhas e multiplicamos a frequência de cada um deles para forçar a existência de duplicatas no dataset de testes.

* **Intervalo 1 (linhas 0 a 2):** Repetido 3 vezes
* **Intervalo 2 (linhas 10 a 14):** Repetido 5 vezes
* **Intervalo 3 (linhas 20 a 22):** Repetido 7 vezes

```python
import pandas as pd
df = pd.read_csv("netflix_titles.csv")

df_modificado = df.copy()

intervalo1 = df_modificado.iloc[0:3]
intervalo2 = df_modificado.iloc[10:15]
intervalo3 = df_modificado.iloc[20:23]

df_modificado = pd.concat([df] + [intervalo1] * 2 + [intervalo2] * 3 + [intervalo3] * 7, ignore_index=True,)
```

#### 8.2 Apagando Células na Coluna 'release_year'

A coluna `release_year` é a única variável numérica do dataset e encontra-se originalmente 100% preenchida. Para podermos testar a funcionalidade do menu, selecionamos aleatoriamente algumas linhas e removemos seus valores (transformando-os em `NaN`).

```python
# Quantidade de linhas a ser removida
qtd_numeros_removidos = 100

# Usando a função 'sample' para selecionar índices aleatórios no DataFrame
indices_aleatorios = df_modificado.sample(n=qtd_numeros_removidos, random_state=42).index

# Substitui o ano de lançamento dessas linhas por valores nulos (NaN)
df_modificado.loc[indices_aleatorios, "release_year"] = None


print(f"Sucesso! {qtd_numeros_removidos} valores foram removidos de 'release_year'.")
print(f"Total de valores vazios na coluna 'release_year': {df_modificado['release_year'].isnull().sum()}")
```

```python
#Verificando valores nulos na coluna 'release_year'
df_modificado.isnull().sum()
```

#### 8.3 Salvando o Dataset Modificado

Após a inserção das duplicatas e a remoção de valores na coluna `release_year`, exportamos o novo conjunto de dados em formato CSV. 

Utilizamos o parâmetro `index=False` para evitar a criação de uma coluna adicional com os índices do Pandas.


```python
# Salva o arquivo CSV sem a coluna de índice do Pandas
df_modificado.to_csv("netflix_titles_modificado.csv", index=False)
```
